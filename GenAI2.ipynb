```python
def call_claude_conversation(messages, model=None, max_tokens=256, temperature=0.5, system="You are a helpful assistant."): 
    """ 
    Call Claude API for multi-turn conversations. 
     
    Args: 
        messages (list): List of message dicts with 'role' and 'content' 
        model (str): Model to use (defaults to MODEL_NAME from .env) 
        max_tokens (int): Maximum tokens to generate 
        temperature (float): Creativity level 
        system (str): System prompt 
     
    Returns: 
        dict: Response with text and metadata 
    """     
    import requests 
    from dotenv import load_dotenv
    import os 
    load_dotenv(override=True)

    API_KEY = os.getenv("ANTHROPIC_API_KEY") 
    ENDPOINT = os.getenv("BASE_URL") + "/v1/messages" 
    MODEL_NAME = os.getenv("MODEL_NAME")

    if model is None: 
        model = MODEL_NAME 
     
    headers = { 
        "x-api-key": API_KEY, 
        "anthropic-version": "2023-06-01", 
        "content-type": "application/json" 
    } 
     
    payload = { 
        "model": model, 
        "max_tokens": max_tokens, 
        "temperature": temperature, 
        "system": system, 
        "messages": messages 
    } 
     
    response = requests.post(ENDPOINT, headers=headers, json=payload) 
     
    if response.status_code == 200: 
        data = response.json() 
        return { 
            "text": data["content"][0]["text"], 
            "status": "success" 
        } 
    else: 
        return { 
            "error": response.status_code, 
            "status": "error" 
        } 
 
# ─── Example multi-turn conversation ────────────────────────────────────────── 
print("=" * 70) 
print("Multi-turn Conversation Example") 
print("=" * 70) 
 
conversation = [ 
    { 
        "role": "user", 
        "content": "What is Python?" 
    } 
] 
 
# First turn 
print("\n[User] What is Python?") 
result = call_claude_conversation(conversation, max_tokens=300) 
if result["status"] == "success": 
    print(f"\n[Claude] {result['text']}") 
    # Add to conversation history 
    conversation.append({"role": "assistant", "content": result["text"]}) 
 
 
# Second turn 
print("\n" + "-" * 70) 
#conversation.append({"role": "user", "content": "What are its main uses?"}) 
print("\n[User] What are its main uses?") 
result = call_claude_conversation(conversation, max_tokens=300) 
if result["status"] == "success": 
    print(f"\n[Claude] {result['text']}") 
    conversation.append({"role": "assistant", "content": result["text"]}) 
 
 
# Third turn 
print("\n" + "-" * 70) 
conversation.append({"role": "user", "content": "How do I get started?"}) 
print("\n[User] How do I get started?") 
result = call_claude_conversation(conversation, max_tokens=300) 
if result["status"] == "success": 
    print(f"\n[Claude] {result['text']}") 
```
